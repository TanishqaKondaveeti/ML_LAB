import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset from sklearn
iris_raw = load_iris()


iris_df = pd.DataFrame(data=iris_raw.data, columns=iris_raw.feature_names)

# Map numeric targets to taxonomic species names
iris_df['species_id'] = iris_raw.target
species_mapping = {i: name for i, name in enumerate(iris_raw.target_names)}
iris_df['species_name'] = iris_df['species_id'].map(species_mapping)

print("Iris dimensions:", iris_df.shape)
print("Missing entries checking:\n", iris_df.isnull().sum())
print(iris_df.head(3))


# Calculating central tendency across variables
for col in iris_raw.feature_names:
    mean_val = iris_df[col].mean()
    median_val = iris_df[col].median()
    mode_val = iris_df[col].mode()[0]
    print(f"Feature: {col}")
    print(f"  Mean:   {mean_val:.4f} cm")
    print(f"  Median: {median_val:.4f} cm")
    print(f"  Mode:   {mode_val:.4f} cm")
    print("-" * 40)


# Calculating dispersion characteristics
for col in iris_raw.feature_names:
    variance = iris_df[col].var()
    std_dev = iris_df[col].std()
    data_range = iris_df[col].max() - iris_df[col].min()
    q1 = iris_df[col].quantile(0.25)
    q3 = iris_df[col].quantile(0.75)
    print(f"Feature: {col}")
    print(f"  Variance:           {variance:.4f}")
    print(f"  Std Deviation:      {std_dev:.4f}")
    print(f"  Range (Max - Min):  {data_range:.4f}")
    print(f"  IQR (Q3 - Q1):      {q3 - q1:.4f}")
    print("-" * 40)

# Evaluating distribution moments
for col in iris_raw.feature_names:
    skewness = iris_df[col].skew()
    kurtosis = iris_df[col].kurt()
    print(f"Feature: {col}")
    print(f"  Skewness (Asymmetry): {skewness:.4f}")
    print(f"  Kurtosis (Tails):     {kurtosis:.4f}")
    print("-" * 40)

```python
# Segmenting by class species using groupby
grouped_species = iris_df.groupby('species_name')

print("--- Group-Wise Mean Values ---")
print(grouped_species[iris_raw.feature_names].mean())

print("\n--- Group-Wise Standard Deviation Profiles ---")
print(grouped_species[iris_raw.feature_names].std())


aggregated_stats = iris_df.groupby('species_name')['petal length (cm)'].agg(['mean', 'median', 'std', 'skew'])
print("Aggregated Statistics (Petal Length):\n", aggregated_stats)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris Dataset
df = sns.load_dataset("iris")

# Display first five rows
print(df.head())
```

```python
# Histogram
plt.figure(figsize=(6,4))
sns.histplot(df["sepal_length"], bins=15, kde=True)
plt.title("Histogram of Sepal Length(24EU02022)")
plt.show()

# Count Plot
plt.figure(figsize=(6,4))
sns.countplot(x="species", data=df)
plt.title("Species Count(24EU02022)")
plt.show()

# Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(y="petal_length", data=df)
plt.title("Box Plot of Petal Length(24EU02022)")
plt.show()

# Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(y="petal_width", data=df)
plt.title("Violin Plot of Petal Width(24EU02022)")
plt.show()

# Density Plot
plt.figure(figsize=(6,4))
sns.kdeplot(df["sepal_width"], fill=True)
plt.title("Density Plot of Sepal Width(24EU02022)")
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(x="sepal_length", y="petal_length",
                hue="species", data=df)
plt.title("Sepal Length vs Petal Length(24EU02022)")
plt.show()

# Pair Plot
sns.pairplot(df, hue="species")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(7,5))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap(24EU02022)")
plt.show()

# Box Plot by Species
plt.figure(figsize=(6,4))
sns.boxplot(x="species", y="petal_length", data=df)
plt.title("Petal Length by Species")
plt.show()

# Violin Plot by Species
plt.figure(figsize=(6,4))
sns.violinplot(x="species", y="sepal_width", data=df)
plt.title("Sepal Width by Species")
plt.show()

# Bar Plot
plt.figure(figsize=(6,4))
sns.barplot(x="species", y="petal_width", data=df)
plt.title("Average Petal Width by Species")
plt.show()
```

```python
# ============================================================
# Import Libraries
# ============================================================
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

import warnings
warnings.filterwarnings("ignore")

# ============================================================
# Load Dataset
# ============================================================
titanic = sns.load_dataset("titanic")

print("Dataset Preview:")
display(titanic.sample(5))

# ============================================================
# Basic Information
# ============================================================
print("Shape of Dataset:", titanic.shape)

print("\nColumns:")
print(list(titanic.columns))

print("\nData Types:")
print(titanic.dtypes)

print("\nDataset Info:")
titanic.info()

# ============================================================
# Descriptive Statistics
# ============================================================
print("\nNumerical Summary")
display(titanic.describe(include='all'))

# ============================================================
# Missing Values Analysis
# ============================================================
missing = titanic.isnull().sum()

print("\nMissing Values:")
print(missing)

print("\nMissing Value Percentage:")
print(round((missing / len(titanic)) * 100, 2))

print("\nDuplicate Records:", titanic.duplicated().sum())

# ============================================================
# Univariate Analysis
# ============================================================

# Survival Distribution
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='survived', palette='Set2')
plt.title("Passenger Survival")
plt.show()

# Passenger Class
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='pclass', palette='pastel')
plt.title("Passenger Class")
plt.show()

# Gender
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='sex', palette='husl')
plt.title("Gender Distribution")
plt.show()

# Age
plt.figure(figsize=(8,5))
sns.histplot(titanic['age'], bins=25, color='skyblue', kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()

# Fare
plt.figure(figsize=(8,5))
sns.histplot(titanic['fare'], bins=35, color='orange', kde=True)
plt.title("Fare Distribution")
plt.show()

# Boxplots
fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.boxplot(y=titanic['age'], ax=ax[0], color='lightgreen')
ax[0].set_title("Age Boxplot")

sns.boxplot(y=titanic['fare'], ax=ax[1], color='salmon')
ax[1].set_title("Fare Boxplot")

plt.show()

# ============================================================
# Bivariate Analysis
# ============================================================

# Gender vs Survival
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='sex', hue='survived', palette='viridis')
plt.title("Gender vs Survival")
plt.show()

# Passenger Class vs Survival
plt.figure(figsize=(6,4))
sns.barplot(data=titanic, x='pclass', y='survived', palette='coolwarm')
plt.title("Survival Rate by Passenger Class")
plt.show()

# Age vs Fare
plt.figure(figsize=(8,5))
sns.scatterplot(data=titanic,
                x='age',
                y='fare',
                hue='survived',
                style='sex')
plt.title("Age vs Fare")
plt.show()

# ============================================================
# Correlation
# ============================================================

numeric_cols = titanic.select_dtypes(include=np.number)

corr = numeric_cols.corr()

print("\nCorrelation Matrix:")
display(corr)

plt.figure(figsize=(9,7))
sns.heatmap(corr,
            annot=True,
            cmap="YlGnBu",
            fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

# Pair Plot
sns.pairplot(
    titanic[['survived','age','fare','pclass']],
    hue='survived',
    diag_kind='hist'
)
plt.show()

# ============================================================
# Category Counts
# ============================================================

print("\nGender Counts")
print(titanic['sex'].value_counts())

print("\nPassenger Class Counts")
print(titanic['pclass'].value_counts())

print("\nEmbarked Counts")
print(titanic['embarked'].value_counts())

# ============================================================
# Data Cleaning
# ============================================================

# Fill Missing Values
titanic['age'] = titanic['age'].fillna(titanic['age'].mean())

titanic['embarked'] = titanic['embarked'].fillna(
    titanic['embarked'].mode()[0]
)

titanic['embark_town'] = titanic['embark_town'].fillna(
    titanic['embark_town'].mode()[0]
)

# Remove Deck Column
titanic = titanic.drop(columns=['deck'])

# ============================================================
# Verify Cleaning
# ============================================================

print("\nMissing Values After Cleaning")
print(titanic.isnull().sum())

print("\nFinal Dataset Shape:", titanic.shape)

print("\nRandom Sample After Cleaning")
display(titanic.sample(5))
```
